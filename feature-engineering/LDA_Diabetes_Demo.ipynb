```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
# import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.metrics import confusion_matrix, classification_report, precision_score
```

## Load the Dataset

```python
# we will be using a diabetes dataset with no missing values (ref: KNN Imputer)
data = pd.read_csv('diabetes_imputed.csv')
X = data.drop('Outcome',axis=1)
y = data[:]['Outcome']

colnames = list(data) #good way to extract column names from dataframe
```

```python
colnames
```

```python
X.head(10)
```

# Split Train and Test

```python
# No need for standardizaion in LDA
# Split Training & Testing datasets
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3, random_state=123)
```

# Apply the LDA

```python
lda = LDA()
X_train = lda.fit(X_train, y_train)

# Percentage of class labels (0 or 1)
print("Model Priors:\n",lda.priors_)
```

# Coefficients

```python
lda.coef_
```

```python
# Binary classification will result in 1 Component only
lda_coef = pd.DataFrame(np.transpose(lda.coef_), index=list(X))
print("\nLDA Coefficients:\n",lda_coef)
```

# Apply LDA on the Test set

```python
X_test
```

```python
lda.predict(X_test)
```

```python
y_test
```

```python
pred=lda.predict(X_test)
print(np.unique(pred, return_counts=True))
confmatrix = pd.DataFrame(confusion_matrix(pred, y_test, labels=[0,1]), index=[0,1])
print("\nConfusion Matrix:\n", confmatrix)
print('\n')
print(classification_report(y_test, pred, digits=3))
```

# Using LDA to predict new data

```python
# say we have a new case and we want to predict diabetes
newpatient = [6,148,72,35,0,33.6,0.627,50]
diagnosis = lda.predict([newpatient])
print('Predicted Class: %d' % diagnosis)
print('Predicted Class: ', diagnosis)
```
