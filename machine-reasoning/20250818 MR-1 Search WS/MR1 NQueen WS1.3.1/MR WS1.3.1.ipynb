```python
from ortools.constraint_solver import pywrapcp

# Define the size of the chessboard
N = 8

# Create the solver object
solver = pywrapcp.Solver('n-queens')

# Create decision variables: one IntVar per column, representing the row position of a queen
queens = [solver.IntVar(0, N - 1, f'x{i}') for i in range(N)]

# Add constraints:
# 1️⃣ All queens must be in different rows
solver.Add(solver.AllDifferent(queens))

# 2️⃣ No two queens can be on the same diagonal
solver.Add(solver.AllDifferent([queens[i] + i for i in range(N)]))
solver.Add(solver.AllDifferent([queens[i] - i for i in range(N)]))

# Define search strategy (how the solver explores solutions)
db = solver.Phase(queens,
                  solver.CHOOSE_FIRST_UNBOUND,
                  solver.ASSIGN_MIN_VALUE)

# Start the search
solver.NewSearch(db)

# Iterate through all solutions
while solver.NextSolution():
    print([int(queens[i].Value()) for i in range(N)])

# End the search
solver.EndSearch()

# Print statistics
print("Solutions found:", solver.Solutions())
print("Failures:", solver.Failures())
print("Branches:", solver.Branches())
print("Wall time:", solver.WallTime(), "ms")
```

```python
import sys
from ortools.constraint_solver import pywrapcp


def main(board_size):
    # Creates the solver.
    solver = pywrapcp.Solver("n-queens")

    # Creates the variables.
    # The array index is the column, and the value is the row.
    queens = [solver.IntVar(0, board_size - 1, f"x{i}") for i in range(board_size)]

    # Creates the constraints.
    # All rows must be different.
    solver.Add(solver.AllDifferent(queens))

    # No two queens can be on the same diagonal.
    solver.Add(solver.AllDifferent([queens[i] + i for i in range(board_size)]))
    solver.Add(solver.AllDifferent([queens[i] - i for i in range(board_size)]))

    db = solver.Phase(queens, solver.CHOOSE_FIRST_UNBOUND, solver.ASSIGN_MIN_VALUE)

    # Iterates through the solutions, displaying each.
    # effectively printing a visual chessboard for each solution found
    num_solutions = 0
    solver.NewSearch(db)
    while solver.NextSolution():
        # Displays the solution just computed.
        for i in range(board_size):
            for j in range(board_size):
                if queens[j].Value() == i:   # write your own code
                    # There is a queen in column j, row i.
                    print('Q', end=' ')  # write your own code
                else:
                    # Empty cell
                    print('.', end=' ') # write your own code
            print() # new line after each row
        print() # blank line between solutions
        num_solutions += 1
    solver.EndSearch()

    # Statistics.
    print("\nStatistics")
    print(f"  failures: {solver.Failures()}")
    print(f"  branches: {solver.Branches()}")
    print(f"  wall time: {solver.WallTime()} ms")
    print(f"  Solutions found: {num_solutions}")


if __name__ == "__main__":
    # By default, solve the 8x8 problem.
    size = 6
    # if len(sys.argv) > 1:
    #     size = int(sys.argv[1])
    main(size)
```
