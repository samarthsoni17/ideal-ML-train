Here, I use inverse transform sampling to generate random samples from a Exponential distribution, step by step.
rate λ=2.

The CDF is:
F(x)=1−e^(−λx)

The inverse (PPF) is: 
F^(−1)(u)= −1/λ * ln(1−u)

```python
import numpy as np

# Parameters
lam = 2
n_samples = 10000

# Step 1: Generate uniform random numbers in [0, 1] -- these will be a set of probabilities to help us get actual samples
u = np.random.rand(n_samples)

# Step 2: Apply the inverse CDF (PPF) -- get actual samples from the distribution by mapping random probabilities to the PPF
x = -np.log(1 - u) / lam
# x now contains samples from the exponential distribution with rate lam
```

Visual Check using plotly

```python
import plotly.graph_objects as go

# Histogram of sampled data (normalized to probability density)
# 10000 samples reduced to 50 bins
hist_chart1 = go.Histogram(
    x=x,
    histnorm='probability density',  # so area under bars ~1
    nbinsx=50,
    name='Sampled',
    opacity=0.7
)

# Theoretical PDF
x_vals = np.linspace(0, 4, 200) #200 random numbers between 0 and 4 would be enough to visualise the exponential distribution
pdf = lam * np.exp(-lam * x_vals)
expdistpdf_chart2 = go.Scatter(
    x=x_vals,
    y=pdf,
    mode='lines',
    name='Theoretical PDF',
    line=dict(color='red')
)

# Combine the two charts and plot
fig = go.Figure([hist_chart1, expdistpdf_chart2])
fig.update_layout(
    title='Inverse Transform Sampling: Exponential Distribution',
    xaxis_title='x',
    yaxis_title='Density',
    bargap=0.1
)
fig.show()
```
