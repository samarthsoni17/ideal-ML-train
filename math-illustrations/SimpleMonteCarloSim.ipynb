# Simple monte carlo
This is a simple demonstration, will be extended later


```python
import numpy as np
numsims = 1000000 #1 M simulations

# assume time to complete each report can vary normally
A_report_time = np.random.uniform(1, 5, numsims)
B_report_time = np.random.uniform(2, 6, numsims)

total_completion_time = A_report_time + B_report_time
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize = (3,1.5))
plt.hist(total_completion_time, density = True, bins = 20)
plt.axvline(9, color = 'red') #show the mean
plt.show()
```

```python
print("probability of unable to complete both reports in 9 hours: " + str((total_completion_time > 9).sum()/numsims))
```
